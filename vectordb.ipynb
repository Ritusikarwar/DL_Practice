```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
from langchain_core.documents import Document
```

```python
doc1=Document(page_content="This is the first document.", metadata={"source": "doc1"})
doc2=Document(page_content="This is the second document.", metadata={"source": "doc2"})
docs=[doc1, doc2]
```

```python
pip install sentence-transformers
```

```python
embedding_model=HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
vector_store=Chroma(
    embedding_function=embedding_model,
    persist_directory="./chroma_db",
    collection_name="my_collection"
)
```

```python
vector_store.add_documents(docs)
```

```python
vector_store.get(include=['embeddings', 'metadatas', 'documents'])
```

```python
vector_store.similarity_search(query="What is the content of the first document?",k=1)
```

```python
vector_store.similarity_search_with_score(query="What is the content of the first document?",k=1)
```

```python
vector_store.similarity_search(query="",filter={"source": "doc1"},k=1)
```
