```python
corpus = "Hey there! how are you? Life has been a lot easier lately. Seeing you at the park yesterday made me feel the storm of emotions i have not felt in years. It's been 5 years Sarah i hope you are doing alright. I know you will throw this letter as well in trash can but if the angel guards are by my side i hope you read this one"

splitter = CharacterTextSplitter(
    chunk_size=10,
    chunk_overlap=0,
    separator=''
)

chunks = splitter.split_text(corpus)

print(chunks)
```

```python
corpus = "Hey there! how are you? Life has been a lot easier to you lately. Seeing you at the park yesterday made me feel the storm of emotions i havenot felt in years. It's been 5 years Sarah i hope you are doing alright. I know you will throw this letter as well in trash can but if the angel guards are by my side i hope you read this one"

splitter = RecursiveCharacterTextSplitter(
    chunk_size=10,
    chunk_overlap=0
)

chunks = splitter.split_text(corpus)

print(chunks)
```
