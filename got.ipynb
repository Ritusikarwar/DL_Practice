```python
import pandas as pd 
import numpy as np
import gensim
import os
import nltk
from nltk import sent_tokenize
from gensim.utils import simple_preprocess
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
story=[]
for filename in os.listdir('D:\\text_data'):
    f=open(os.path.join('D:\\text_data',filename))
    corpus=f.read()
    raw_sent=sent_tokenize(corpus)
    for sent in raw_sent:
        story.append(simple_preprocess(sent))
```
